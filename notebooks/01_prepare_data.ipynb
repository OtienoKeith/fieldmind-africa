# FieldMind Africa — 01 Prepare open data

Runs on Kaggle, Colab, or a normal CPU. It downloads only the public CC-BY-4.0 FarmerChat dataset and creates deterministic train/validation JSONL plus hashes. No paid API or GPU is used.

```python
# Set this after publishing/forking the repository. Leave blank when the notebook is already inside the repo.
REPO_URL = ""
REPO_DIR = "fieldmind-africa"

from pathlib import Path
import os, subprocess
base = Path('/kaggle/working') if Path('/kaggle/working').exists() else Path('/content') if Path('/content').exists() else Path.cwd()
repo = base / REPO_DIR
if not (repo / 'scripts/prepare_dataset.py').exists():
    if not REPO_URL:
        raise RuntimeError('Set REPO_URL or upload/clone the repository first.')
    subprocess.run(['git', 'clone', '--depth', '1', REPO_URL, str(repo)], check=True)
os.chdir(repo)
print('Repository:', repo)
```
